```python
import pandas as pd
data = pd.read_csv('tech.csv')
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(data['Study Hours'], data['Score'])
plt.title('Study Hours vs Score')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```

```python
plt.bar(data['Score'], data['Name'])
plt.title('Score by Name')
plt.show()
```

Zainab performed best

```python
plt.hist(data['SleepHours'])
plt.title('Distribution of SleepHours')
plt.show()
```

Yes most student had enough sleep

```python
countgender = data['Gender'].value_counts()
```

```python
plt.pie(countgender,labels=countgender.index,autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.show()
```

Both are the same

```python
import seaborn as sns
```
